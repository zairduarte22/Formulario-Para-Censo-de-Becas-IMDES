### Realizar las importaciones necesarias para la limpieza

```python
import pandas as pd
import numpy as np
import datetime
```

### Leer y mostrar los primeros registros del csv para estudiar su estructura

```python
data = pd.read_csv('mock_becas.csv')
data.head()
```

### Realizar los reemplazos necesarios para crear graficos eficientemente

```python
data['ubicacion_universidad'] = data['ubicacion_universidad'].replace({'La Villa': 'Villa del Rosario', 'foraneo':'Maracaibo'}, inplace=True)
data['servicio_comunitario'] = data['servicio_comunitario'].replace({'SI':1, 'NO':0}, inplace=True)
data['trabaja'] = data['trabaja'].replace({'SI':1, 'NO':0}, inplace=True)
data.head()
```

### Calcular la fecha de nacimiento para poder categorizar a los estudiantes en menores o mayores de edad

```python
data['fecha_nacimiento'] = pd.to_datetime(data['fecha_nacimiento'], errors='coerce')
data['edad'] = (datetime.datetime.now() - data['fecha_nacimiento']).dt.days // 365
# Creación de la columna "edad_categorica" basada en la edad (Binning)
# limite = 18
"""data["categoria_edad"] = np.where(
    data["edad"] < limite,
    "Menor de 18 años",
    "Mayor o igual a 18 años"
)"""
# Creación de la columna "edad_categorica" usando pd.cut() para categorizar la edad en "Menor de edad" y "Mayor de edad"
bins_edad = [-np.inf, 17, np.inf]
labels_edad = ['Menor de edad', 'Mayor de edad']
data['edad_categorica'] = pd.cut(data['edad'], bins=bins_edad, labels=labels_edad, include_lowest=True)
data.head()
```

### Finalmente, se exporta el csv con las columnas necesarias para construir el dashboard con las metricas solicitadas.

```python
data.to_csv('mock_becas_limpio.csv', index=False)
```
